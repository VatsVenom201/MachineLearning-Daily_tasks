```python
import numpy as np
import pandas as pd
```

```python
x_train = pd.read_csv('Data/X_train.csv')
x_test = pd.read_csv('Data/X_test.csv')
y_train = pd.read_csv('Data/y_train.csv')
y_test = pd.read_csv('Data/y_test.csv')

x_train.head()
```

```python
df1 = pd.concat([x_train, x_test], ignore_index=True)
df1.head(10)
```

```python
df2 = pd.concat([y_train,y_test],ignore_index=True)
df2.head()
```

```python
df = pd.merge(df1,df2,on='filghtId')
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder, StandardScaler, OneHotEncoder
```

```python
le = LabelEncoder()
stdsc = StandardScaler()
ohe = OneHotEncoder()
```

```python
df.head(10)
```

```python
del df['filghtId']
del df['flight']
df.head(10)
```

```python
df['airline'].value_counts()
```

```python
# removing rare categories

threshold = 50  # minimum rows to keep
def category_filter(df,col):
    
    counts = df[col].value_counts()
    valid_categories = counts[counts >= threshold].index

    df = df[df[col].isin(valid_categories)]
    return df
df = category_filter(df,'airline')
df = category_filter(df,'departure_time')
df['airline'].value_counts()
```

```python
columns = ['airline','source_city','departure_time','stops','arrival_time','destination_city']

for col in columns:
    df[col] = le.fit_transform(df[col])
     
    print(f"\nColumn: {col}")
    print("Mapping:", dict(zip(le.classes_, le.transform(le.classes_))))
    print(df[col].value_counts())
    
```

```python
df.head()
```

```python
df.info
```

### Scaling

```python
columns = ['duration','days_left']

for col in columns:
    df[col] = stdsc.fit_transform(df[[col]])
df.head()    
```

```python
X = df.drop('price',axis=1)
y = df['price']

X.head()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.model_selection import KFold, cross_val_score
```

## Using KFold Cross Validation we compare 4 diff models

```python
kf = KFold(n_splits=5,shuffle=True,random_state=42)
```

```python
def kcross(model,X,y,cv):

    r2_scores = cross_val_score(model, X, y, cv=cv,scoring='r2')
    
    print(f'Model Used:{model}')
    print(f'R2 Scores:{r2_scores}')
    print(f'Mean R2:{r2_scores.mean()}')
    print(f'Standard Deviation R2:{r2_scores.std()}')
```

```python
kcross(LinearRegression(),X,y,kf)
```

```python
kcross(SVR(kernel='rbf'),X,y,kf)
```

```python
rf = RandomForestRegressor(
    n_estimators=300,
    max_depth=None,
    random_state=42,
    n_jobs=-1
)
kcross(rf,X,y,kf)
```

```python
xgb = XGBRegressor(
    n_estimators=300,
    learning_rate=0.1,
    max_depth=6,
    subsample=0.8,
    random_state=42,
    objective='reg:squarederror',
    n_jobs=-1,
    tree_method='hist'
)

kcross(xgb,X,y,kf)
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test, y_train,y_test = train_test_split(X,y,test_size=0.20)
```

### Through K-Fold Cross Validation we get Best model RandomForestRegressor

```python
from sklearn.model_selection import GridSearchCV
```

```python
rf = RandomForestRegressor(random_state=42)
```

## HyperParameter Tuning using GridSearchCV

```python
param_grid = {
    'n_estimators': [100, 200, 300],           # number of trees
    'max_depth': [None, 10, 20, 30],          # max depth of trees
    'min_samples_split': [2, 5, 10],          # min samples to split a node
    'min_samples_leaf': [1, 2, 4],            # min samples at leaf node
    'max_features': ['auto', 'sqrt', 0.5],    # number of features for split
    'bootstrap': [True]                # use bootstrap samples
}
```

```python
grid = GridSearchCV(estimator=rf,
                    param_grid=param_grid,
                    cv=5,                  # 5-fold cross-validation
                    scoring='r2',  # regression metric
                    n_jobs=-1,             # use all cores
                    verbose=2)

```

```python
grid.fit(X_train,y_train)
```

```python
best_param = grid.best_params_
best_model = grid.best_estimator_

print(f'Best Model:{best_model}')
print(f'Best Parameters:{best_param}')
```

```python
y_pred = best_model.predict(X_test)
```

```python
from sklearn.metrics import r2_score as r2
from sklearn.metrics import root_mean_squared_error as rmse
from sklearn.metrics import mean_squared_error as mse
from sklearn.metrics import mean_absolute_error as mae
```

```python
print(f'R2:{r2(y_test,y_pred)}')
print(f'RMSE:{rmse(y_test,y_pred)}')
print(f'MSE:{mse(y_test,y_pred)}')
print(f'MAE:{mae(y_test,y_pred)}')
```
